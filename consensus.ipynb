```python
from typing import TypedDict, List, Optional
from datetime import datetime

# 1. State Definition
class AgentTrace(TypedDict):
    actor: str
    action: str
    decision: str
    timestamp: str

class ConsensusState(TypedDict):
    task: str
    code_proposal: Optional[str]
    verifier_passed: bool
    arbiter_verdict: Optional[str]
    traces: List[AgentTrace]

# 2. Initial State
state: ConsensusState = {
    "task": "Refactor vector offset calculations for coordinate rendering",
    "code_proposal": "def compute_offset(stroke: int, offset: int = 3): return offset * 2",
    "verifier_passed": False,
    "arbiter_verdict": None,
    "traces": []
}

def log_trace(state: ConsensusState, actor: str, action: str, decision: str):
    state["traces"].append({
        "actor": actor,
        "action": action,
        "decision": decision,
        "timestamp": datetime.now().isoformat()
    })

# 3. Simulate Deterministic Static Verifier Node
if "offset * 2" in (state["code_proposal"] or ""):
    state["verifier_passed"] = True
    log_trace(state, "StaticVerifier", "syntax_and_rule_check", "PASSED")
else:
    state["verifier_passed"] = False
    log_trace(state, "StaticVerifier", "syntax_and_rule_check", "FAILED")

# 4. Simulate Arbiter Gate
if state["verifier_passed"]:
    state["arbiter_verdict"] = "PROCEED_TO_HUMAN_CHECKPOINT"
    log_trace(state, "ModelArbiter", "policy_eval", "APPROVED")
else:
    state["arbiter_verdict"] = "REJECTED_NEEDS_REVISION"
    log_trace(state, "ModelArbiter", "policy_eval", "REJECTED")

print(f"Task: {state['task']}")
print(f"Verifier Status: {state['verifier_passed']}")
print(f"Arbiter Verdict: {state['arbiter_verdict']}")
print("\n--- Telemetry Traces ---")
for t in state["traces"]:
    print(f"[{t['timestamp']}] {t['actor']} -> {t['decision']}")
```
